1. Load the dataset and handle any missing data by replacing null values with the mean
value of the respective column.

```python
import pandas as pd
import numpy as np

dataSet = pd.read_csv("winequality-red.csv", sep =";")
display(dataSet.head())

print(dataSet.isnull().sum())

dataSet = dataSet.fillna(dataSet.mean(numeric_only=True))
display(dataSet) 

print("Missing values after filling:")
print(dataSet.isnull().sum())
```

2. Extract the following columns as vectors: alcohol,citric acid.

```python
alcohol_vector = dataSet["alcohol"].to_numpy()
citric_vector = dataSet["citric acid"].to_numpy()
print("Alcohol Vector:" ,alcohol_vector)
print("Citric Acid Vector:",citric_vector)
```

3. Select two features (e.g., alcohol and density) from the dataset and calculate the
covariance matrix using np.cov(X.T), where X is the feature matrix consisting of the
selected columns.

```python
X = dataSet[["alcohol", "density"]].to_numpy()
covariance_matrix = np.cov(X.T)
print("Covariance Matrix: \n", covariance_matrix)
```

4. Perform eigen decomposition on the covariance matrix,results:Identify the top 2 eigenvalues,Identify the corresponding eigenvectors.

```python
eigenValues, eigenVectors = np.linalg.eigh(covariance_matrix)
indices = np.argsort(eigenValues)[::-1]
eigenValues = eigenValues[indices]
eigenVectors = eigenVectors[:, indices]
print("Top 2 Eigenvalues:")
print(eigenValues[:2])
```

5. Most common wine quality in the dataset and interpret distribution of wine quality scores.

```python
qualityScore_count = dataSet["quality"].value_counts().sort_index()
print("The distribution of wine quality:",qualityScore_count)

mostCommon_quality = dataSet["quality"].mode()[0]
mostCommon_count = dataSet["quality"].value_counts().max()

print("Most common wine quality:", mostCommon_quality)
print("Number of wines:", mostCommon_count)

percentage = (mostCommon_count / len(dataSet)) * 100

roundPercent = percentage.round(2)

print("Percentage:",roundPercent)
```
